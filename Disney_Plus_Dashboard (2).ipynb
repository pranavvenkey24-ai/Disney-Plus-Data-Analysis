**Import libraries and load dataset**

```python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go
import numpy as np

# Load dataset
file_name = next(iter(uploaded))
df = pd.read_csv(file_name)

print("Dataset loaded successfully!")
```

**Exploratory Data Analysis (EDA)**

```python
# Display dataset shape
print("Dataset Shape:", df.shape)

# Display first five records
print("\nFirst 5 Records:")
display(df.head())

# Dataset information
print("\nDataset Information:")
df.info()

# Check missing values
print("\nMissing Values:")
display(df.isnull().sum())
```

**Data Cleaning and Preprocessing**

```python
# Create a copy of the original dataset
clean_df = df.copy()

# Convert date_added into datetime format
clean_df["date_added"] = pd.to_datetime(
    clean_df["date_added"],
    errors="coerce"
)

# Handle missing values
clean_df["director"] = clean_df["director"].fillna("Unknown")
clean_df["cast"] = clean_df["cast"].fillna("Unknown")
clean_df["country"] = clean_df["country"].fillna("Unknown")
clean_df["rating"] = clean_df["rating"].fillna("Not Rated")

# Display cleaned dataset
print("Cleaned Dataset Shape:", clean_df.shape)

print("\nRemaining Missing Values:")
display(clean_df.isnull().sum())
```

**Data Visualisation**

**Chart 1 — Bar Chart: Movies vs TV Shows**

```python
# Count Movies and TV Shows

content_data = df["type"].value_counts().reset_index()

content_data.columns = ["Content Type", "Count"]

# Display the data
display(content_data)
```

```python
fig1 = px.bar(
    content_data,
    x="Content Type",
    y="Count",
    color="Content Type",
    title="Movies vs TV Shows on Disney+",
    text="Count",
    template="plotly_dark"
)

fig1.update_layout(
    title_x=0.5,
    height=500
)

fig1.show()
```

**Chart 2 — Line Chart**

```python
# Group titles by release year and count them

yearly_data = (
    clean_df.groupby("release_year")
    .size()
    .reset_index(name="Total Titles")
)

# Sort by release year
yearly_data = yearly_data.sort_values("release_year")

# Display the data
display(yearly_data.head(10))
```

```python
fig2 = px.line(
    yearly_data,
    x="release_year",
    y="Total Titles",
    markers=True,
    title="Disney+ Content Releases Over the Years",
    labels={
        "release_year": "Release Year",
        "Total Titles": "Number of Titles"
    },
    template="plotly_dark"
)

fig2.update_layout(
    title_x=0.5,
    height=500
)

fig2.show()
```

**Chart 3 — Scatter Plot**

```python
# Select only movies
movie_data = clean_df[clean_df["type"] == "Movie"].copy()

# Extract duration in minutes
movie_data["duration_minutes"] = (
    movie_data["duration"]
    .str.extract(r"(\d+)")
    .astype(float)
)

# Display prepared data
display(movie_data[
    ["title", "release_year", "duration_minutes"]
].head(10))
```

```python
fig3 = px.scatter(
    movie_data,
    x="release_year",
    y="duration_minutes",
    title="Movie Release Year vs Duration",
    labels={
        "release_year": "Release Year",
        "duration_minutes": "Duration (Minutes)"
    },
    template="plotly_dark"
)

fig3.update_layout(
    title_x=0.5,
    height=500
)

fig3.show()
```

**Chart 4 — Histogram: Movie Duration Distribution**

```python
# Select only movies
movie_duration = clean_df[clean_df["type"] == "Movie"].copy()

# Extract duration in minutes
movie_duration["duration_minutes"] = (
    movie_duration["duration"]
    .str.extract(r"(\d+)")
    .astype(float)
)

# Display movie durations
display(movie_duration[["title", "duration_minutes"]].head(10))
```

```python
fig4 = px.histogram(
    movie_duration,
    x="duration_minutes",
    nbins=20,
    title="Distribution of Movie Durations",
    labels={
        "duration_minutes": "Movie Duration (Minutes)"
    },
    color_discrete_sequence=["#636EFA"],
    template="plotly_dark"
)

fig4.update_layout(
    title_x=0.5,
    height=500,
    yaxis_title="Number of Movies"
)

fig4.show()
```

**Chart 5 — Heatmap: Content Type vs Release Decade**

```python
# Create release decade from release year
clean_df["Decade"] = (
    (clean_df["release_year"] // 10) * 10
)

# Group data by content type and decade
heatmap_data = clean_df.groupby(
    ["type", "Decade"]
).size().unstack(fill_value=0)

# Display prepared data
display(heatmap_data)
```

```python
fig5 = px.imshow(
    heatmap_data,
    text_auto=True,
    color_continuous_scale="Blues",
    title="Movies and TV Shows Released by Decade",
    labels={
        "x": "Release Decade",
        "y": "Content Type",
        "color": "Number of Titles"
    },
    template="plotly_dark"
)

fig5.update_layout(
    title_x=0.5,
    height=500
)

fig5.show()
```

```python

import ipywidgets as widgets
from IPython.display import display, clear_output

# Prepare dashboard data
dashboard_df = clean_df.copy()

# Extract duration values
dashboard_df["duration_value"] = (
    dashboard_df["duration"]
    .str.extract(r"(\d+)")
    .astype(float)
)

# Create release decade
dashboard_df["Decade"] = (
    dashboard_df["release_year"] // 10
) * 10


# Create filters
type_filter = widgets.Dropdown(
    options=["All", "Movie", "TV Show"],
    value="All",
    description="Content Type:"
)

year_filter = widgets.IntRangeSlider(
    value=(
        int(dashboard_df["release_year"].min()),
        int(dashboard_df["release_year"].max())
    ),
    min=int(dashboard_df["release_year"].min()),
    max=int(dashboard_df["release_year"].max()),
    description="Year Range:",
    continuous_update=False
)


# Create chart output areas
out1 = widgets.Output()
out2 = widgets.Output()
out3 = widgets.Output()
out4 = widgets.Output()
out5 = widgets.Output()

total_titles = widgets.HTML()


# Dashboard function
def update_dashboard(change=None):

    # Apply filters
    filtered_df = dashboard_df[
        dashboard_df["release_year"].between(
            year_filter.value[0],
            year_filter.value[1]
        )
    ]

    if type_filter.value != "All":
        filtered_df = filtered_df[
            filtered_df["type"] == type_filter.value
        ]

    total_titles.value = (
        f"<h3>Total Titles: {len(filtered_df)}</h3>"
    )


    # Chart 1: Bar Chart
    with out1:
        clear_output(wait=True)

        content_data = (
            filtered_df["type"].value_counts().reset_index()
        )

        content_data.columns = ["Content Type", "Count"]

        fig1 = px.bar(
            content_data,
            x="Content Type",
            y="Count",
            color="Content Type",
            title="Movies vs TV Shows",
            template="plotly_dark",
            height=350
        )

        display(fig1)


    # Chart 2: Line Chart
    with out2:
        clear_output(wait=True)

        yearly_data = (
            filtered_df.groupby("release_year")
            .size()
            .reset_index(name="Count")
        )

        fig2 = px.line(
            yearly_data,
            x="release_year",
            y="Count",
            markers=True,
            title="Content Releases Over the Years",
            template="plotly_dark",
            height=350
        )

        display(fig2)


    # Prepare duration data
    if type_filter.value == "TV Show":

        duration_data = filtered_df[
            filtered_df["type"] == "TV Show"
        ].copy()

        duration_label = "Number of Seasons"

    else:

        duration_data = filtered_df[
            filtered_df["type"] == "Movie"
        ].copy()

        duration_label = "Duration (Minutes)"


    duration_data = duration_data.dropna(
        subset=["duration_value"]
    )


    # Chart 3: Scatter Plot
    with out3:
        clear_output(wait=True)

        if not duration_data.empty:

            if type_filter.value == "TV Show":
                chart_title = "TV Show Seasons vs Release Year"
            elif type_filter.value == "Movie":
                chart_title = "Movie Duration vs Release Year"
            else:
                chart_title = "Movie Duration vs Release Year (Movies Only)"

            fig3 = px.scatter(
                duration_data,
                x="release_year",
                y="duration_value",
                title=chart_title,
                labels={
                    "release_year": "Release Year",
                    "duration_value": duration_label
                },
                template="plotly_dark",
                height=350
            )

            display(fig3)


    # Chart 4: Histogram
    with out4:
        clear_output(wait=True)

        if not duration_data.empty:

            if type_filter.value == "TV Show":
                histogram_title = "Distribution of TV Show Seasons"
            elif type_filter.value == "Movie":
                histogram_title = "Distribution of Movie Durations"
            else:
                histogram_title = "Distribution of Movie Durations (Movies Only)"

            fig4 = px.histogram(
                duration_data,
                x="duration_value",
                nbins=20,
                title=histogram_title,
                labels={
                    "duration_value": duration_label
                },
                template="plotly_dark",
                height=350
            )

            display(fig4)


    # Chart 5: Heatmap
    with out5:
        clear_output(wait=True)

        heatmap_data = filtered_df.groupby(
            ["type", "Decade"]
        ).size().unstack(fill_value=0)

        if not heatmap_data.empty:

            fig5 = px.imshow(
                heatmap_data,
                text_auto=True,
                color_continuous_scale="Blues",
                title="Content Type vs Release Decade",
                template="plotly_dark",
                height=350
            )

            display(fig5)


# Connect filters
type_filter.observe(update_dashboard, names="value")
year_filter.observe(update_dashboard, names="value")


# Arrange dashboard in two columns
dashboard_layout = widgets.GridspecLayout(
    3, 2,
    height="1150px"
)

dashboard_layout[0, 0] = out1
dashboard_layout[0, 1] = out2
dashboard_layout[1, 0] = out3
dashboard_layout[1, 1] = out4
dashboard_layout[2, :] = out5


# Display dashboard
display(
    widgets.HTML(
        "<h1>Disney+ Content Analysis Dashboard</h1>"
    ),
    widgets.HBox([type_filter, year_filter]),
    total_titles,
    dashboard_layout
)

# Display initial charts
update_dashboard()
```

## Conclusion

The Disney+ dataset was analysed to understand its content distribution and release trends.

### Key Findings
- The dataset contains **1,450 titles**.
- Movies account for **1,052 titles**, while TV Shows account for **398 titles**.
- Movies make up approximately **72.6%** of the dataset, showing that movies form the majority of the available content.
- The year **2021** recorded the highest number of titles by release year, with 125 titles.
- Interactive charts were developed to explore the dataset based on content type and release year.

### Limitations
- Some information, such as director, cast and country, was missing in the original dataset.
- Three missing `date_added` values were retained rather than replaced with assumed dates.
- The analysis is limited to the information available in the dataset.
